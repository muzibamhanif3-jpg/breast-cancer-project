# **OUTLIER DETECTION**

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df[["Tumor Size", "Survival Months"]], orient="h")

plt.title("Outlier Detection for Tumor Size and Survival Months")
plt.show()
```

```python
#IQR METHOD:

Q1 = df["Tumor Size"].quantile(0.25)
Q3 = df["Tumor Size"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Tumor Size"] < lower_bound) |
    (df["Tumor Size"] > upper_bound)
]

outliers.shape
```
